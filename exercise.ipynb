```python
#  Exercise 

# Using functions, conditions, error handling, and input:

# 1. Create a simple banking app that can accept deposits, withdrawals, and display the balance.

# 2. Create an app that calculates the area of a circle and a rectangle (the user should choose the shape and enter the required measurements).

# 3. Write a function that checks the strength of a password

import math
def banking_app():
    balance = 0.0

    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print(f"Deposited: Ksh{amount:.2f}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(amount):
        nonlocal balance
        if 0 < amount <= balance:
            balance -= amount
            print(f"Withdrew: ${amount:.2f}")
        else:
            print("Insufficient funds or invalid withdrawal amount.")

    def display_balance():
        print(f"Current balance: Ksh{balance:.2f}")

    while True:
        action = input("Choose an action: deposit, withdraw, balance, exit: ").strip().lower()
        if action == "deposit":
            try:
                amount = float(input("Enter deposit amount: "))
                deposit(amount)
            except ValueError:
                print("Invalid input. Please enter a numeric value.")
        elif action == "withdraw":
            try:
                amount = float(input("Enter withdrawal amount: "))
                withdraw(amount)
            except ValueError:
                print("Invalid input. Please enter a numeric value.")
        elif action == "balance":
            display_balance()
        elif action == "exit":
            print("Exiting banking app.")
            break
        else:
            print("Invalid action. Please choose again.")
banking_app()

```

```python
#area calculator

def area_calculator():
    def circle_area(radius):
        return 3.142 * radius * radius
    def rectangle_area(length, width):
        return length * width
    while True:
        shape = input("Choose a shape (circle/rectangle) or 'exit' to quit: ").strip().lower()
        if shape == "circle":
            try:
                radius = float(input("Enter the radius of the circle: "))
                area = circle_area(radius)
                print(f"Area of the circle: {area:.2f}")
            except ValueError:
                print("Invalid input. Please enter a numeric value.")
        elif shape == "rectangle":
            try:
                length = float(input("Enter the length of the rectangle: "))
                width = float(input("Enter the width of the rectangle: "))
                area = rectangle_area(length, width)
                print(f"Area of the rectangle: {area:.2f}")
            except ValueError:
                print("Invalid input. Please enter numeric values.")
        elif shape == "exit":
            print("Exiting area calculator.")
            break
        else:
            print("Invalid shape. Please choose again.")
area_calculator()
```

```python
#password strength checker
def password_strength_checker(password):
    if len(password) < 8:
        return "Weak: Password must be at least 8 characters long."
    if not any(char.isdigit() for char in password):
        return "Weak: Password must contain at least one digit."
    if not any(char.isupper() for char in password):
        return "Weak: Password must contain at least one uppercase letter."
    if not any(char.islower() for char in password):
        return "Weak: Password must contain at least one lowercase letter."
    if not any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/`~" for char in password):
        return "Weak: Password must contain at least one special character."
    return "Strong password."
user_password = input("Enter a password to check its strength: ")
strength = password_strength_checker(user_password)
print(user_password)
password_strength_checker(user_password)
```

```python
#implementation using dictionaries
def calculate_average( assignment, cat, exam):
    return (assignment + cat + exam)/3
def assign_letter_grade(average):
    if average >= 90:
        return 'A'
    elif average >= 80:
        return 'B'
    elif average >= 70:
        return 'C'
    elif average >= 60:
        return 'D'
    else:
        return 'F'
students = {}
while True:
    name = input("Enter student name (or 'exit' to finish): ")
    if name.lower() == 'exit':
        break
    try:
        assignment = float(input("Enter assignment score: "))
        cat = float(input("Enter CAT score: "))
        exam = float(input("Enter exam score: "))
    except ValueError:
        print("Invalid input. Please enter numeric values for scores.")
        continue
    average = calculate_average(assignment, cat, exam)
    letter_grade = assign_letter_grade(average)
    students[name] = {'average': average, 'grade': letter_grade}
    
for name, info in students.items():
    print(f"Student: {name}, Average: {info['average']:.2f}, Grade: {info['grade']}")


```
