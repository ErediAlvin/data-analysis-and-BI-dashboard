# One Sample t-test(Two-Tailed)

**Step 1: State the hypotheses**
 
$H_0: \mu = 75$  vs  $H_1: \mu \neq 75$
 
**Step 2: Select appropriate test**
 
- One sample (Final grade)
- Population standard deviation is unknown
- Sample size is relatively small
- Thus the appropriate test to use 1-sample t-test
 
**Step 3: Choose significance level, alpha = 0.05**
 
- We will use the p-value approach
 
**Step 4: Compute the test statistic - p-value**

```python
t_stat, p_value = stats.ttest_1samp(df["FinalGrade"], 75,alternative='two-sided')
 
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

**Step 5:Compare the p-value with alpha and make decision**

```python
alpha = 0.05
 
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

**Step 6: Conclusion**
- At 5% significance level, we do not have sufficient evidence that the average final grade differs from 75.Based on this sample, we cannot confidently say the class is performing better or worse than the benchmark of 75.

# One Sample t-test (Upper-Tail test)
 **Someone else comes about and state that they think the mean is more than 75**
 **Step 1: State the hypotheses**
 
$H_0: \mu = 75$  vs  $H_1: \mu > 75$
 
**Step 2: Select appropriate test**
 
- One sample (Final grade)
- Population standard deviation is unknown
- Sample size is relatively small
- Thus the appropriate test to use 1-sample t-test
 
**Step 3: Choose significance level, alpha = 0.05**
 
- We will use the p-value approach
 
**Step 4: Compute the test statistic - p-value**

```python
t_stat, p_value = stats.ttest_1samp(df["FinalGrade"], 75,alternative='greater')
 
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

**Step 5:Compare the p-value with alpha and make decision**

```python
alpha = 0.05
 
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

**Step 6: Conclusion**
 
- At the 5% significance level, we do not have sufficient evidence that the average final grade is greater than 75. Based on this sample, we cannot confidently conclude that the class is performing above the benchmark of 75.
