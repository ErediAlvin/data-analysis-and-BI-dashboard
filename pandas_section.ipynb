```python
# Question: Create 4 separate lists of your own for a meeting. Then combine them into a dictionary. Then convert into a pandas df.

attendees = ["Alice", "Bob", "Charlie", "David"]
times = ["10:00 AM", "10:15 AM", "10:30 AM", "10:45 AM"]
dates = ["2026-07-02", "2026-09-02", "2026-06-02", "2026-07-02"]
locations = ["Room 101", "Room 102", "Room 103", "Room 104"]


meeting_dict = {
    'Attendee': attendees,  
    'Time': times,
    'Date': dates,  
    'Location': locations
}
meeting_df = pd.DataFrame(meeting_dict)
meeting_df
```

```python
std_df2 = pd.DataFrame(meeting_dict, index=['A', 'B', 'C', 'D'])
std_df2
```

#### Creating a df from a list of Dictionaries


```python
# creating a list of dictionaries where each dictionary represents a row of data using the premier league teams
premier_league_teams = [
    {"Team": "Manchester United", "Points": 55, "Position": 3},
    {"Team": "Liverpool", "Points": 80, "Position": 1},
    {"Team": "Chelsea", "Points": 75, "Position": 2},
    {"Team": "Arsenal", "Points": 60, "Position": 4},
    {"Team": "Manchester City", "Points": 70, "Position": 5}
]
premier_league_df = pd.DataFrame(premier_league_teams)
premier_league_df
#sort by points
premier_league_df.sort_values(by='Points', ascending=False)
```
