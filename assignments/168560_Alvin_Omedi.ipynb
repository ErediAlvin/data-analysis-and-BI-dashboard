### Question 16
Write a Python function that takes a list of numbers and returns a new list with only the prime numbers.

```python
# output prime numbers after taking an input from user
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
user_input = int(input("Enter a number to find all prime numbers up to that number: "))
print(f"Prime numbers up to {user_input}:")
for n in range(2, user_input + 1):
    if is_prime(n):
        print(n)
        
```

Create a Python program that simulates a student course registration system. The system should allow the following:

```python
# Student Course Enrollment System
# Features:
#1.Register a new student with a unique ID, name, and email.
#2.Add available courses (e.g., Python, Data Science, Web Development).
#3.Allow students to enroll in one or more courses.
#4.Display all students and their enrolled courses.
#5.Allow updating a student's email.
#6.Allow removing a student or a course.
#7.Use functions for each operation.
#8.Use dictionaries to store student data and lists for courses.
#9.Use loops to allow multiple operations until the user chooses to exit.

# Use input functions to interact with the user.
# Use dictionaries and lists to manage data.
# Use functions to organize your code.
# Use loops and conditionals to control the flow.
# Ensure the program handles invalid input gracefully.



def register_student(students):
    student_id = input("Enter student ID: ")
    if student_id in students:
        print("Student ID already exists. Please try again.")
        return
    name = input("Enter student name: ")
    email = input("Enter student email: ")
    students[student_id] = {"name": name, "email": email, "courses": []}
    print(f"Student {name} registered successfully.")

def add_course(courses):
    course_name = input("Enter course name: ")
    if course_name in courses:
        print("Course already exists. Please try again.")
        return
    courses.append(course_name)
    print(f"Course {course_name} added successfully.")

def enroll_student(students, courses):
    student_id = input("Enter student ID: ")
    if student_id not in students:
        print("Student ID not found. Please try again.")
        return
    course_name = input("Enter course name to enroll: ")
    if course_name not in courses:
        print("Course not found. Please try again.")
        return
    students[student_id]["courses"].append(course_name)
    print(f"Student {students[student_id]['name']} enrolled in {course_name} successfully.")

def display_students(students):
    for student_id, info in students.items():
        print(f"ID: STD-{student_id}, Name: {info['name']}, Email: {info['email']}, Courses: {', '.join(info['courses'])}")

def update_email(students):
    student_id = input("Enter student ID: ")
    if student_id not in students:
        print("Student ID not found. Please try again.")
        return
    new_email = input("Enter new email: ")
    students[student_id]["email"] = new_email
    print(f"Email for student {students[student_id]['name']} updated successfully.")
def remove_student(students):
    student_id = input("Enter student ID to remove: ")
    if student_id not in students:
        print("Student ID not found. Please try again.")
        return
    del students[student_id]
    print(f"Student with ID {student_id} removed successfully.")
def remove_course(courses):
    course_name = input("Enter course name to remove: ")
    if course_name not in courses:
        print("Course not found. Please try again.")
        return
    courses.remove(course_name)
    print(f"Course {course_name} removed successfully.")


def main():
    students = {}
    courses = []
    while True:
        print("\nStudent Course Enrollment System")
        print("1. Register a new student")
        print("2. Add a new course")
        print("3. Enroll a student in a course")
        print("4. Display all students and their enrolled courses")
        print("5. Update a student's email")
        print("6. Remove a student")
        print("7. Remove a course")
        print("8. Exit")
        choice = input("Enter your choice: ")
        print("="* 50)
        
        if choice == '1':
            register_student(students)
        elif choice == '2':
            add_course(courses)
        elif choice == '3':
            enroll_student(students, courses)
        elif choice == '4':
            display_students(students)
        elif choice == '5':
            update_email(students)
        elif choice == '6':
            remove_student(students)
        elif choice == '7':
            remove_course(courses)
        elif choice == '8':
            print("Exiting the system. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    main()
```
