#### 3.Data Analysis

3.1 Univariate analysis

```python
df.describe()[['finalWorth', 'age', 'life_expectancy_country']]
```

```python
# Histogram of te age
sns.histplot(data=df, x='age', bins=30, kde=True)
sns.despine()
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# checking the skewness of the age distribution
skewness = df['age'].skew()
print(f'Skewness of Age Distribution: {skewness}')
```

```python
#interpretation of the skewness value
if skewness > 0:
    print("The age distribution is positively skewed (right-skewed).")
elif skewness < 0:
    print("The age distribution is negatively skewed (left-skewed).")   
```

```python
df['age'].plot(kind='box')
plt.title('Boxplot of Age')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

```python
# distribution by gender

df['gender'].unique()
```

```python
sns.set_theme(style='whitegrid')
sns.countplot(data=df, x='gender')
sns.despine()
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')   

#downloading the image
plt.savefig('gender_distribution.png', dpi=300, bbox_inches='tight')  
      
plt.show()
```

### insight


```python
# distribution of the countries with the most billionaires
top_countries = df['country'].value_counts().nlargest(10)
sns.set_theme(style='whitegrid')
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
sns.despine()
plt.title('Top 10 Countries with the Most Billionaires')
plt.xlabel('Number of Billionaires')
plt.ylabel('Country')
plt.show
```
