```python
class Calculator:

    def __init__(self, num1, num2):
        """Pass Arg . 2 Numbers num1 and num2"""
        self.num1 = num1
        self.num2 = num2

    def adds(self):
        """Adds two given numbers"""
        return self.num1 + self.num2

    def subtract(self):
        """Subtracts the second number from the first"""
        return self.num1 - self.num2

    def division_(self):
        """Divides the first number by the second"""
        if self.num2 == 0:
            print("You can't divide a number by zero")
        else:
            return self.num1 / self.num2
```

```python
data = df.groupby('cut', observed=True).agg({'carat': 'mean'}).reset_index()
data
```

```python
fig , ax = plt. subplots()
ax.barh(data['cut'], data['carat'])
fig.show()
```

```python
fig , ax = plt. subplots()
ax.barh(data['cut'],data['carat'])
ax.set_title("Carat Vs Cut", fontsize=14, fontweight='bold')
ax.set_xlabel("Carat", fontsize=12)
ax.set_ylabel("Cut", fontsize=12)
ax.grid(True, alpha=0.3, linestyle='--')
fig.tight_layout()
fig.savefig("part2b_oop_style.png", dpi=120)
fig.show()
```

```python
flights = sns.load_dataset("flights")

pivot = flights.pivot(index="year", columns="month", values="passengers")

fig, ax = plt.subplots(figsize=(8,5))
ax.plot(pivot.index, pivot["Jan"])
ax.set_title("Line Chart – January Flights Over Years")
ax.set_xlabel("Year")
ax.set_ylabel("Passengers")
plt.show()
```

```python
tips    = sns.load_dataset("tips")
avg_bill = tips.groupby("day",observed=True)["total_bill"].mean()

fig, ax = plt.subplots()
ax.bar(avg_bill.index, avg_bill.values)
ax.set_title("Average Bill per Day")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.barh(avg_bill.index, avg_bill.values)
ax.set_title("Average Bill per Day (Horizontal)")
plt.show()
```

```python
import numpy as np

grouped = tips.groupby(["day", "sex"])["total_bill"].mean().unstack()

x = np.arange(len(grouped.index))
width = 0.35

fig, ax = plt.subplots()
ax.bar(x - width/2, grouped["Male"], width, label="Male")
ax.bar(x + width/2, grouped["Female"], width, label="Female")

ax.set_xticks(x)
ax.set_xticklabels(grouped.index)
ax.legend()
ax.set_title("Grouped Bar Chart")
plt.show()
```

```python
iris = sns.load_dataset("iris")

fig, ax = plt.subplots()
ax.scatter(iris["sepal_length"], iris["sepal_width"])
ax.set_title("Sepal Length vs Width")
plt.show()
```

```python
fig, ax = plt.subplots()
sizes = iris["petal_length"] * 20

ax.scatter(
    iris["sepal_length"],
    iris["sepal_width"],
    s=sizes,
    alpha=0.5
)

ax.set_title("Bubble Chart")
plt.show()
```

```python
penguins = sns.load_dataset("penguins")

fig, ax = plt.subplots()
ax.hist(penguins["flipper_length_mm"].dropna(), bins=15)
ax.set_title("Histogram – Flipper Length")
plt.show()
```
