```python
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/NUELBUNDI/Machine-Learning-Data-Set/refs/heads/main/Instagram-Data.csv')
```

```python
df.head()
```

```python
rows, columns = df.shape
print(f'The dataset has {rows} rows and {columns} columns.')
df.info()
```

```python
df.sample(5)
```

```python
df['Audience Country'].unique()
```

```python
audience_country_counts = df['Audience Country'].value_counts()
plt.figure(figsize=(13, 6))
audience_country_counts.plot(kind='bar')
plt.title('Distribution of Audience Country')
plt.xlabel('Audience Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()  
```
